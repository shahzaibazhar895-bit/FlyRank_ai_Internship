## 3. Success metric

Success Metric


The primary success metric will be Precision@50.

Precision@50 measures how many of the top 50 pages selected by the system are actually declining according to the chosen proxy label.

A good result is one that identifies a high proportion of relevant pages within the top-ranked recommendations because reviewers can only review a limited number of pages at a time.

## 5. Why ML beats a fixed rule here

Why ML Beats a Fixed Rule Here


A fixed rule may only consider a few conditions, such as whether a page is old or has high impressions.

Machine learning can evaluate multiple signals simultaneously, including CTR, impressions, engagement, content age, freshness, and trend information.

This allows the system to identify more complex patterns that are difficult to capture with a single hand-written rule. The goal is to support better prioritization decisions, not replace human judgment.
